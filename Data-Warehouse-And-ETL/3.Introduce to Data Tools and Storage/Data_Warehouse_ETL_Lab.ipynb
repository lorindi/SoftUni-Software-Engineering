## 5. Read a CSV file directly from AWS S3

Install `s3fs`, then use `pd.read_csv()` with an `s3://bucket/path` URL to load data straight from cloud storage — no manual download step needed.

```
# pip install s3fs

AWS_ACCESS_KEY = os.getenv("AWS_ACCESS_KEY")
AWS_SECRET_KEY = os.getenv("AWS_SECRET_KEY")

df3 = pd.read_csv(
    "s3://engineering-course-9/IntroDataToolsLab/sales.csv",
    storage_options={
        "key": AWS_ACCESS_KEY,
        "secret": AWS_SECRET_KEY,
        "client_kwargs": {"region_name": "eu-north-1"}
    }
)
print(df3)
```

## 12. Merge two DataFrames

Join sales with customer data on a shared key — like a SQL JOIN, but in Python.

```
print(df3.columns)

df4 = pd.read_csv(
    "s3://engineering-course-9/IntroDataToolsLab/customers.csv",
    storage_options={
        "key": AWS_ACCESS_KEY,
        "secret": AWS_SECRET_KEY,
        "client_kwargs": {"region_name": "eu-north-1"}
    }
)
print(df4.columns)
print("-----------------------------------------------------------")
merged_df = pd.merge(df3, df4, on="customer_id")
print(merged_df.head(10))


```

## 13. Save a DataFrame to PostgreSQL

Use `DataFrame.to_sql()` to persist your data into a real database table.

```
import os
from dotenv import load_dotenv

load_dotenv(override=True)

# --- AWS credentials ---
AWS_ACCESS_KEY = os.getenv("AWS_ACCESS_KEY")
AWS_SECRET_KEY = os.getenv("AWS_SECRET_KEY")

# --- PostgreSQL credentials ---
DB_HOST = os.getenv("DB_HOST")
DB_USER = os.getenv("DB_USER")
DB_PASSWORD = os.getenv("DB_PASSWORD")
DB_PORT = os.getenv("DB_PORT")
DB_NAME = os.getenv("DB_NAME")
```

```
from sqlalchemy import create_engine
from sqlalchemy.engine import URL

url = URL.create(
    drivername="postgresql+psycopg2",
    username=DB_USER,
    password=DB_PASSWORD,
    host=DB_HOST,
    port=int(DB_PORT),
    database=DB_NAME
)

engine = create_engine(url)
merged_df.to_sql("merged_sales", engine, index=False, if_exists="replace")
```
