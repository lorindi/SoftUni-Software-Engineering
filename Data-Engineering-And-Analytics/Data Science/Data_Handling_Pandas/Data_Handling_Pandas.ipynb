* When you write df_fruits["Fruit"], you tell Pandas: "Give me only this column." Pandas pulls the column from the table and returns it as a Series (one dimension).
* When you write df_fruits[["Fruit"]], you say: "Give me a small table (subset) with these columns."
Because you asked for a table (the list inside double brackets), Pandas returns a DataFrame (two dimensions), even if there is only one column.

```python
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv" 
df_iris = pd.read_csv(url)
```

```python
df_iris
```

```python
df_iris.info()
```

```python
df_iris.describe()
```

```python
df_iris["sepal_length"].max()
```

```python
df_iris[((df_iris["species"] == "virginica") & (df_iris["petal_length"] > 6.0))]
```

```python
df_iris.groupby("species")["petal_length"].mean()
```

```python
df_iris["species"].value_counts()
```

```python
df_iris_filter =  df_iris[df_iris["sepal_width"] > 3.0] 
df_iris_filter.groupby("species")["petal_length"].mean()
```

```python
df_iris_setosa = df_iris[df_iris["species"] == "setosa"]
df_iris_setosa
```

```python
df_iris_setosa["petal_length"].sum()
```

```python
# 1. Build a mapping dictionary for readable labels
translate_dict = { "setosa": "Setosa", "versicolor": "Versicolor", "virginica": "Virginica" } 
```

```python
# 2. Add the new column with .map()
df_iris["species_en"] = df_iris["species"].map(translate_dict)
```

```python
# 3. Check the result (first 5 rows)
df_iris[["species", "species_en"]].head()
```

```python
df_iris[["species", "species_en"]].sample(10)
```

```python
df_iris.groupby("species")["petal_length"].mean()
```

```python
all_means = df_iris.groupby("species").mean() 
all_means
```

```python
all_means = df_iris.groupby("species").mean(numeric_only = True) 
all_means
```

```python
iris_stats = df_iris.groupby("species")["petal_length"].agg(["mean", "min", "max"]) 
iris_stats
```

```python
iris_stats = df_iris.groupby("species")["petal_length"].agg(["mean", "min", "max", "sum", "count"]) 
iris_stats
```

```python
df_iris.groupby("species_en")["petal_length"].mean()
```

```python
df_iris["flower_length"] = 	df_iris["petal_length"]
df_iris.groupby(["species", "flower_length"])["petal_length"].mean()
```
